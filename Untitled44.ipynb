```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
FILE_PATH = "Amazon Sale Report.csv"     # change path if needed
OUT_DIR = "eda_charts"
os.makedirs(OUT_DIR, exist_ok=True)
sns.set_theme(style="whitegrid", palette="viridis")
plt.rcParams.update({"figure.dpi": 110, "axes.titlesize": 13, "axes.titleweight": "bold"})
```

```python
def save(name):
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR, name), bbox_inches="tight")
    plt.show()
```

```python
df = pd.read_csv(FILE_PATH, low_memory=False)
print("Raw shape:", df.shape)
```

```python
df
```

```python
# 2.1 Column names: strip spaces, lowercase, replace - and spaces with _
df.columns = (df.columns.str.strip().str.lower()
              .str.replace(r"[\s\-]+", "_", regex=True))
df.columns
```

```python
# 2.2 Drop useless columns (empty junk column + redundant index)
df = df.drop(columns=[c for c in ["unnamed:_22", "index"] if c in df.columns])
df
```

```python
# 2.3 Remove duplicate rows
before = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print(f"Duplicates removed: {before - len(df)}")
```

```python
# 2.4 Fix data types
df["date"] = pd.to_datetime(df["date"], format="%m-%d-%y", errors="coerce")
df["ship_postal_code"] = df["ship_postal_code"].astype("Int64").astype("string")
df["b2b"] = df["b2b"].astype(bool)

print(df.dtypes)
print("\nInvalid dates:", df["date"].isna().sum())
print("Date range:", df["date"].min(), "to", df["date"].max())
df.head()
```

```python
df["ship_postal_code"] = pd.to_numeric(df["ship_postal_code"], errors="coerce").astype("Int64").astype("string")
```

```python
# 2.5 Standardise text columns
for col in ["ship_city", "ship_state"]:
    print(f"{col}: {df[col].nunique()} unique values, {df[col].isna().sum()} missing")

print("\nTop states:")
print(df["ship_state"].value_counts().head(10))

df[["ship_city", "ship_state"]].head()
    
```

```python
state_map = {
    "RAJSHTHAN": "Rajasthan", "RAJSTHAN": "Rajasthan", "RJ": "Rajasthan",
    "PUNJAB/MOHALI/ZIRAKPUR": "Punjab", "PB": "Punjab",
    "NEW DELHI": "Delhi",
    "ORISSA": "Odisha",
    "PONDICHERRY": "Puducherry",
    "NL": "Nagaland",
    "AR": "Arunachal Pradesh",
    "DADRA AND NAGAR": "Dadra and Nagar Haveli",
    "APO": pd.NA,  # military post office, not a real state
}
df["ship_state"] = df["ship_state"].str.upper()
df["ship_state"] = df["ship_state"].replace(state_map)
df["ship_state"] = df["ship_state"].str.title()
```

```python
# Preview cleaned data
print(df[["ship_city", "ship_state"]].head())

# All unique states after cleaning
print("\nUnique states:")
print(sorted(df["ship_state"].dropna().unique()))

# Count per state
print("\nOrders per state:")
print(df["ship_state"].value_counts(dropna=False))
```

```python
print(df["ship_state"].value_counts().tail(20))
```

```python
def status_group(s):
    if pd.isna(s):
        return "Unknown"
    if s == "Cancelled":
        return "Cancelled"
    if "Returned" in s or "Returning" in s or "Rejected" in s:
        return "Returned"
    if s.startswith("Pending"):
        return "Pending"
    return "Shipped/Delivered"

df["status_group"] = df["status"].apply(status_group)

# 2.7 Handle missing values
df["promotion_ids"] = df["promotion_ids"].fillna("None")
df["fulfilled_by"] = df["fulfilled_by"].fillna("Not Easy Ship")
df["courier_status"] = df["courier_status"].fillna("Unknown")
df["currency"] = df["currency"].fillna("INR")           # only INR exists in data
df["ship_country"] = df["ship_country"].fillna("IN")    # only IN exists in data
df["has_promo"] = df["promotion_ids"].ne("None")
df["amount"] = df["amount"].astype(float)
```

```python
# Status groups
print("Status group counts:")
print(df["status_group"].value_counts(dropna=False))

# Missing values remaining (should be 0 for the filled columns)
print("\nMissing values:")
print(df[["promotion_ids", "fulfilled_by", "courier_status",
          "currency", "ship_country", "amount"]].isna().sum())

# Promo flag
print("\nOrders with promo:")
print(df["has_promo"].value_counts())

# Amount column check
print("\nAmount dtype:", df["amount"].dtype)
print(df["amount"].describe())

# Preview
print(df[["status", "status_group", "promotion_ids", "has_promo", "amount"]].head(10))
```

```python
df.loc[df["status_group"].eq("Cancelled") & df["amount"].isna(), "amount"] = 0
```

```python
# Cancelled orders should now have no missing amounts
print("Cancelled with NaN amount:",
      (df["status_group"].eq("Cancelled") & df["amount"].isna()).sum())

# Remaining missing amounts (non-cancelled)
print("Total missing amounts:", df["amount"].isna().sum())

# Which rows are still missing?
print("\nMissing by status group:")
print(df.loc[df["amount"].isna(), "status_group"].value_counts())

# Amount summary by status group
print("\nAmount summary by status group:")
print(df.groupby("status_group")["amount"].agg(["count", "sum", "mean"]).round(2))

# Preview the rows that were just set to 0
print(df.loc[df["status_group"].eq("Cancelled"), ["status", "status_group", "amount"]].head())
```

```python
# 2.8 Add helper columns for analysis
df["month"] = df["date"].dt.to_period("M").dt.to_timestamp()
df["weekday"] = df["date"].dt.day_name()
df["day"] = df["date"].dt.date
```

```python
# Preview helper columns
print(df[["date", "month", "weekday", "day"]].head(10))

# Date range of the data
print("\nDate range:", df["date"].min(), "to", df["date"].max())

# Orders per month
print("\nOrders per month:")
print(df["month"].value_counts().sort_index())

# Orders per weekday (in calendar order)
order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
print("\nOrders per weekday:")
print(df["weekday"].value_counts().reindex(order))

# Revenue per month (excluding cancelled)
print("\nRevenue per month (non-cancelled):")
print(df[df["status_group"] != "Cancelled"].groupby("month")["amount"].sum().round(2))

# Missing dates check
print("\nMissing dates:", df["date"].isna().sum())
```

```python
# 2.9 Order size sanity: Qty > 5 looks like a data-entry error; flag it
df["qty_outlier"] = df["qty"] > 5
```

```python
# "valid" = real sales (not cancelled, has an amount)
valid = df[(df["status_group"] != "Cancelled") & df["amount"].notna() & (df["amount"] > 0)]

print("Clean shape:", df.shape)
print("Remaining missing values:\n", df.isna().sum()[df.isna().sum() > 0], "\n")
df.to_csv("amazon_sales_clean.csv", index=False)
```

```python
# 3.1 Daily revenue trend with 7-day rolling average
daily = valid.groupby("day")["amount"].sum()
daily.index = pd.to_datetime(daily.index)
daily = daily[daily.index >= "2022-04-01"]  # March has only a few days of data
plt.figure(figsize=(12, 5))
plt.plot(daily.index, daily.values, alpha=0.4, label="Daily")
plt.plot(daily.index, daily.rolling(7).mean(), linewidth=2.5, label="7-day average")
plt.title("Daily Revenue (INR)")
plt.legend()
save("01_daily_revenue")
```

```python
# 3.2 Monthly revenue
monthly = valid.groupby("month")["amount"].sum()
plt.figure(figsize=(8, 5))
ax = sns.barplot(x=monthly.index.strftime("%b %Y"), y=monthly.values)
ax.bar_label(ax.containers[0], fmt="%.0f", fontsize=8)
plt.title("Monthly Revenue (INR) - March is a partial month")
plt.xlabel("")
save("02_monthly_revenue")
```

```python
# 3.3 Revenue by category
cat = valid.groupby("category")["amount"].sum().sort_values()
plt.figure(figsize=(9, 5))
cat.plot(kind="barh", color=sns.color_palette("deep")[0])
plt.title("Revenue by Category (INR)")
plt.ylabel("")
save("03_revenue_by_category")
```

```python
# 3.4 Top 10 states
states = valid.groupby("ship_state")["amount"].sum().nlargest(10).sort_values()
plt.figure(figsize=(9, 5))
states.plot(kind="barh", color=sns.color_palette("deep")[2])
plt.title("Top 10 States by Revenue (INR)")
plt.ylabel("")
save("04_top_states")
```

```python
# 3.5 Top 10 cities
cities = valid.groupby("ship_city")["amount"].sum().nlargest(10).sort_values()
plt.figure(figsize=(9, 5))
cities.plot(kind="barh", color=sns.color_palette("deep")[3])
plt.title("Top 10 Cities by Revenue (INR)")
plt.ylabel("")
save("05_top_cities")
```

```python
# 3.6 Order status breakdown
plt.figure(figsize=(7, 7))
df["status_group"].value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90,
                                        wedgeprops={"width": 0.45}, ylabel="")
plt.title("Order Status")
save("06_order_status")
```

```python
# 3.7 Size distribution (ordered)
size_order = ["XS", "S", "M", "L", "XL", "XXL", "3XL", "4XL", "5XL", "6XL", "Free"]
plt.figure(figsize=(10, 5))
sns.countplot(data=valid, x="size", order=size_order)
plt.title("Units Sold by Size")
plt.xlabel("")
save("07_size_distribution")
```

```python
# 3.8 Fulfilment type: order count and average value
fig, axes = plt.subplots(1, 2, figsize=(11, 5))
sns.countplot(data=valid, x="fulfilment", ax=axes[0])
axes[0].set_title("Orders by Fulfilment")
sns.boxplot(data=valid, x="fulfilment", y="amount", ax=axes[1])
axes[1].set_title("Order Value by Fulfilment")
save("08_fulfilment")
```

```python
# 3.9 Cancellation rate by category
canc = (df.groupby("category")["status_group"]
          .apply(lambda s: (s == "Cancelled").mean() * 100)
          .sort_values())
plt.figure(figsize=(9, 5))
canc.plot(kind="barh", color="tomato")
plt.title("Cancellation Rate by Category (%)")
plt.ylabel("")
save("09_cancellation_by_category")
```

```python
# 3.10 Order value distribution
plt.figure(figsize=(10, 5))
sns.histplot(valid["amount"], bins=50, kde=True)
plt.title("Order Value Distribution (INR)")
save("10_order_value_distribution")
```

```python
# 3.11 Revenue heatmap: category x month
pivot = valid.pivot_table(index="category", columns="month", values="amount", aggfunc="sum")
pivot.columns = pivot.columns.strftime("%b %Y")
plt.figure(figsize=(9, 5))
sns.heatmap(pivot, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Revenue: Category x Month")
plt.ylabel("")
save("11_category_month_heatmap")
```

```python
# 3.12 Weekday pattern
order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
plt.figure(figsize=(9, 5))
sns.countplot(data=valid, x="weekday", order=order)
plt.title("Orders by Day of Week")
plt.xlabel("")
save("12_weekday_orders")
```
